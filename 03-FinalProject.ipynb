## Abstract

This study investigates whether state-level light pollution is associated with depression rates across the United States. Growing evidence suggests that artificial light at night may disrupt circadian rhythms and melatonin production, potentially contributing to adverse mental health outcomes. To explore this relationship at a population level, we combined three datasets: state-level depression statistics (2022–2023), sky brightness observations from Globe at Night 2024, and anxiety/depression prevalence indicators aggregated by state.

After cleaning and merging the datasets by state, we conducted exploratory data analysis and linear regression to quantify the relationship between sky brightness (measured through Sky Quality Meter readings and limiting magnitude) and depression rates. A permutation test was used to assess whether any observed associations exceeded chance. A multiple regression model incorporating three light pollution indicators, sqm_mean, limiting_mag_mean, and light_obs_count, was also fit to evaluate whether effects persisted when controlling for correlated predictors.

Our primary finding is that states with darker average skies show a slight tendency toward lower depression rates, but the association is weak (R² = 0.08). The correlation matrix further reveals that light pollution variables correlate only modestly with depression outcomes (r ≈ −0.09 to −0.16). These results suggest that light pollution may play a small contributing role in population-level mental health, but is unlikely to be a key factor.

We conclude that while a small relationship exists between nighttime sky brightness and state-level depression rates, the relationship is not strong enough to support causal claims from this observational, aggregated analysis alone. Future work should incorporate individual-level data, control for confounders such as urbanization and healthcare access, and account for seasonal variation in both light exposure and mental health outcomes.

## Research Question

How does the state-level nighttime sky brightness magnitude as measured by Globe at Night datasets correlate with depression prevalence and sleep disturbance rates across U.S. states? 

## Background and Prior Work

Light pollution, the common phrase for regions with excessive artificial lighting at night time, has been a long growing concern in regards to the environment, as well as the psychological well being of the public. Studies have shown that being exposed to artificial light at night can disrupt your circadian rhythm, causing mood swings or poor sleep. Satellite imagery tools such as the [Light Pollution Map](https://lightpollutionmap.app/?lat=45.706179&lng=-86.440430&zoom=5&opacity=50&auroraOpacity=90) and observational datasets like Globe at Night let us see the variances in light pollution by area, and we can see a common trend where the more populated the region is, the more light pollution is present. 

These extensive tooling have been used for environmental research, and are reliable in comparing light pollution levels across the U.S. However, most existing projects or studies try to link these stats to mental health outcomes in humans on a large scale. 

The prevalence of depression in humans varies widely by region. Data from the [Center of Disease Control](https://www.cdc.gov/mmwr/volumes/72/wr/mm7224a1.htm) (CDC) highlights that a majority of adults in the U.S. have reported experiencing depression in one form or another, with notable differences between states. Additional state level stats from the [World Population Review](https://worldpopulationreview.com/state-rankings/mental-health-statistics-by-state) and even [government owned websites](https://catalog.data.gov/dataset/adult-depression-lghc-indicator-627e3) further confirm the differences in severity of depression depending on the region. We have extensive datasets to independently track mental health trends and variances in light pollution, but very few projects try to combine mental health indicators and environmental exposure data. As such, we want to build our project on these existing datasets, and study the effects of state level light pollution levels with depression prevalence data to find out whether regions with higher light pollution also report higher cases of depression. 

Recent peer-reviewed research has also supported this connection. A 2024 systematic review and meta-analysis examining over half a million participants found that both indoor and outdoor artificial light at night were associated with increased risks of depression, suggesting that consistent nighttime brightness can negatively affect mental well-being (Chen et al., 2024). 

References: Chen, M., Zhao, Y., Lu, Q., Ye, Z., Bai, A., Xie, Z., Zhang, D., & Jiang, Y. (2024). Artificial light at night and risk of depression: A systematic review and meta-analysis. Environmental Health and Preventive Medicine, 29, 73. 


## Hypothesis


States with higher levels of light pollution will have more cases of diagnosed depression compared to states with lower light pollution levels. This hypothesis is based on existing research discussed in the background section showing that artificial light at night is associated with disruptions in normal sleep patterns and overall well-being. Since the project compares state-level light-pollution intensity data with reported depression statistics it’s reasonable to expect that areas with consistently higher nighttime brightness may also show higher rates of diagnosed depression. The relationship isn’t assumed to be purely causal but rather an observable association between environmental light exposure and mental-health indicators across regions.

### Indicators of Anxiety or Depression (Kaggle)

This dataset will be manually downloaded from Kaggle and placed in `data/00-raw/indicators-of-anxiety-or-depression.csv`.

Key wrangling goals:
- Load and inspect schema (columns, dtypes, missingness)
- Standardize the state key to `state` for merging
- Reduce data to one row per state to avoid row explosion during merge
- Write cleaned output to `data/02-processed/anxiety_depression_indicators_by_state.csv`

```python
# Step A, load the dataset
kaggle_raw_path = 'data/00-raw/mental-health-by-state.csv'
if not os.path.exists(kaggle_raw_path):
    raise FileNotFoundError(
        'Missing file: data/00-raw/indicators-of-anxiety-or-depression.csv. '
        'Download from Kaggle and place it there, then rerun this cell.'
    )

anxiety_raw_df = pd.read_csv(kaggle_raw_path)
anxiety_raw_df.columns = [c.strip() for c in anxiety_raw_df.columns]

anxiety_raw_df.head()
```

```python
# Step B, show size and basic schema
print('Raw anxiety/depression dataset shape:', anxiety_raw_df.shape)
print('\nColumns:\n', anxiety_raw_df.columns.tolist())
print('\nDtypes:\n', anxiety_raw_df.dtypes)
```

```python
# Step C, inspect missingness and identify likely merge/metric columns
print('Missing values by column:\n', anxiety_raw_df.isna().sum().sort_values(ascending=False).head(15))

cols_lower = {c.lower(): c for c in anxiety_raw_df.columns}
state_col = next((cols_lower[k] for k in ['state', 'location'] if k in cols_lower), None)
value_col = next((cols_lower[k] for k in ['value', 'percent', 'percentage', 'estimate'] if k in cols_lower), None)
indicator_col = next((cols_lower[k] for k in ['indicator', 'measure'] if k in cols_lower), None)

print('\nDetected state column:', state_col)
print('Detected value column:', value_col)
print('Detected indicator column:', indicator_col)

if state_col is None:
    raise ValueError('Could not identify a state/location column in the Kaggle CSV.')
```

```python
# Step D, clean and standardize state key
anxiety_working_df = anxiety_raw_df.copy()
anxiety_working_df['state'] = anxiety_working_df[state_col].astype(str).str.strip()
anxiety_working_df = anxiety_working_df[~anxiety_working_df['state'].isin(['United States', 'US', ''])].copy()

if value_col is not None:
    anxiety_working_df[value_col] = (
        anxiety_working_df[value_col]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.replace('%', '', regex=False)
    )
    anxiety_working_df[value_col] = pd.to_numeric(anxiety_working_df[value_col], errors='coerce')

anxiety_working_df.head()
```

```python
# Step E, aggregate to one row per state for merge-safe granularity
if indicator_col is not None and value_col is not None:
    indicator_text = anxiety_working_df[indicator_col].astype(str).str.lower()
    anxiety_working_df['indicator_key'] = np.select(
        [
            indicator_text.str.contains('anxiety') & indicator_text.str.contains('depress'),
            indicator_text.str.contains('anxiety'),
            indicator_text.str.contains('depress')
        ],
        ['anxiety_or_depression_pct', 'anxiety_pct', 'depression_pct'],
        default='other'
    )

    filtered = anxiety_working_df[anxiety_working_df['indicator_key'] != 'other'].copy()
    if filtered.empty:
        anxiety_state_df = (
            anxiety_working_df.groupby('state', as_index=False)[value_col]
            .mean()
            .rename(columns={value_col: 'anxiety_depression_value_mean'})
        )
    else:
        anxiety_state_df = (
            filtered.groupby(['state', 'indicator_key'])[value_col]
            .mean()
            .unstack('indicator_key')
            .reset_index()
        )
else:
    numeric_cols = anxiety_working_df.select_dtypes(include='number').columns.tolist()
    if len(numeric_cols) == 0:
        raise ValueError('No numeric columns found to aggregate from Kaggle dataset.')
    anxiety_state_df = anxiety_working_df.groupby('state', as_index=False)[numeric_cols].mean()

anxiety_state_df['state'] = (
    anxiety_state_df['state']
    .astype(str)
    .str.strip()
    .str.replace(r'\\s+', ' ', regex=True)
)

print('Cleaned state-level anxiety dataset shape:', anxiety_state_df.shape)
anxiety_state_df.head()
```

```python
# Step C, wrangle to merge-ready state tables

def normalize_state(series):
    return (
        series.astype(str)
        .str.strip()
        .str.replace(r'\\s+', ' ', regex=True)
    )

# Keep only US observations from Globe at Night and extract state name
us_light_pol_df = light_pol_df[light_pol_df['Country'].astype(str).str.contains('United States', na=False)].copy()
us_light_pol_df['state'] = us_light_pol_df['Country'].str.split('United States - ').str[-1]
us_light_pol_df['state'] = normalize_state(us_light_pol_df['state'])

# One row per state for merge-safe granularity
light_state_df = (
    us_light_pol_df.groupby('state', as_index=False)
    .agg(
        light_obs_count=('state', 'size'),
        limiting_mag_mean=('LimitingMag', 'mean'),
        sqm_mean=('SQMReading', 'mean')
    )
)

# Standardize state keys in all tables
for df_name, df_obj in [
    ('depression_df', depression_df),
    ('light_state_df', light_state_df),
    ('anxiety_df', anxiety_df),
]:
    df_obj['state'] = normalize_state(df_obj['state'])
    print(f"{df_name} unique states: {df_obj['state'].nunique()}")

print('\nSample of merge-ready light pollution state table:')
light_state_df.head()
```

```python
# Step D, merge all three datasets on state

def unmatched_states(left_df, right_df):
    return sorted(set(left_df['state']) - set(right_df['state']))

print('States in depression not in light_state:', len(unmatched_states(depression_df, light_state_df)))
print('States in depression not in anxiety:', len(unmatched_states(depression_df, anxiety_df)))

combined_df_3way = (
    depression_df
    .merge(light_state_df, on='state', how='inner')
    .merge(anxiety_df, on='state', how='inner')
)

print('\nMerged shape:', combined_df_3way.shape)
print('Merged unique states:', combined_df_3way['state'].nunique())

combined_df_3way.head()
```

### Analysis 1: Linear Regression of Sky Brightness vs Depression

In this first analysis, we quantify the relationship between state-level sky brightness and depression using a linear model. We use `sqm_mean` (higher values indicate darker skies) to predict depression rate, report effect size (`r`, `R^2`), and run a permutation test to evaluate whether the observed slope is stronger than expected by chance.

To make the result easier to interpret, we also fit a multiple linear regression using available light-pollution features (`sqm_mean`, `limiting_mag_mean`, and `light_obs_count` when present). This lets us compare whether sky-brightness effects remain similar when accounting for additional light-pollution indicators.

```python
# Analysis 1: Linear regression + permutation significance test

analysis_df = combined_df_3way.copy()

# Allow this section to work even if naming differs slightly across cleaned tables
possible_targets = ["DepressionRate_2023_x", "DepressionRate_2022_x"]
target_col = next((c for c in possible_targets if c in analysis_df.columns), None)

if target_col is None:
    raise ValueError(f"No expected depression target column found. Available columns: {analysis_df.columns.tolist()}")

x_col = "sqm_mean"
if x_col not in analysis_df.columns:
    raise ValueError("Expected predictor column `sqm_mean` is missing from combined_df_3way.")

# Simple model data prep
simple_df = analysis_df[[x_col, target_col]].dropna().copy()
x = simple_df[x_col].to_numpy(dtype=float)
y = simple_df[target_col].to_numpy(dtype=float)

# Fit simple linear regression using numpy
slope, intercept = np.polyfit(x, y, 1)
y_hat = slope * x + intercept

# Correlation and explained variance
r = np.corrcoef(x, y)[0, 1]
r2 = r ** 2

print(f"Target variable used: {target_col}")
print(f"Rows used in simple regression: {len(simple_df)}")
print(f"Simple model equation: {target_col} = {intercept:.4f} + ({slope:.4f}) * {x_col}")
print(f"Pearson r: {r:.4f}")
print(f"R^2: {r2:.4f}")

# Permutation test for slope significance
rng = np.random.default_rng(108)
n_perm = 5000
perm_slopes = np.empty(n_perm)
for i in range(n_perm):
    y_perm = rng.permutation(y)
    perm_slopes[i] = np.polyfit(x, y_perm, 1)[0]

p_value = (np.sum(np.abs(perm_slopes) >= abs(slope)) + 1) / (n_perm + 1)
print(f"Permutation-test p-value for slope (two-sided, {n_perm} shuffles): {p_value:.4f}")

# Plot with fitted line
plt.figure(figsize=(8, 6))
sns.scatterplot(data=simple_df, x=x_col, y=target_col, s=70)

x_line = np.linspace(x.min(), x.max(), 200)
y_line = slope * x_line + intercept
plt.plot(x_line, y_line, color="red", linewidth=2, label="Linear fit")

plt.title("First Analysis: Sky Brightness vs Depression (State Level)")
plt.xlabel("Average Sky Brightness (sqm_mean; higher = darker skies)")
plt.ylabel(f"{target_col} (%)")
plt.legend()
plt.show()

# Multiple regression (if additional light-pollution predictors are available)
predictor_candidates = ["sqm_mean", "limiting_mag_mean", "light_obs_count"]
predictors = [c for c in predictor_candidates if c in analysis_df.columns]

multi_df = analysis_df[predictors + [target_col]].dropna().copy()
if len(predictors) >= 2 and len(multi_df) >= 10:
    X_raw = multi_df[predictors].to_numpy(dtype=float)
    y_multi = multi_df[target_col].to_numpy(dtype=float)

    # Standardize predictors so coefficient magnitudes are comparable
    X_mean = X_raw.mean(axis=0)
    X_std = X_raw.std(axis=0, ddof=0)
    X_std[X_std == 0] = 1.0
    X_z = (X_raw - X_mean) / X_std

    X_design = np.column_stack([np.ones(len(X_z)), X_z])
    beta, *_ = np.linalg.lstsq(X_design, y_multi, rcond=None)
    y_multi_hat = X_design @ beta

    ss_res = np.sum((y_multi - y_multi_hat) ** 2)
    ss_tot = np.sum((y_multi - y_multi.mean()) ** 2)
    r2_multi = 1 - (ss_res / ss_tot)

    print("\nMultiple linear regression (standardized predictors):")
    print(f"Rows used: {len(multi_df)}")
    print(f"R^2: {r2_multi:.4f}")
    print(f"Intercept: {beta[0]:.4f}")
    for name, coef in zip(predictors, beta[1:]):
        print(f"Coefficient for {name}: {coef:.4f}")
else:
    print("\nSkipped multiple regression: not enough predictors or non-missing rows.")
```

#### Analysis 1 Interpretation

The first analysis suggests that states with darker average skies (`sqm_mean`) tend to have slightly lower depression rates, but the pattern is weak. On the plot, the trend line has the expected direction, yet many states are spread far from that line, which means sky brightness by itself does not predict depression very strongly.

For our project, this is still an important result. It supports the idea that light pollution may be related to mental-health outcomes, but it also shows that depression rates are likely influenced by many other factors beyond nighttime brightness alone. Overall, Analysis 1 points to a small association rather than a strong statewide effect.
